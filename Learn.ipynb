```python
import ipywidgets as widgets
from IPython.display import display

# Defineerime valemid (kasutame topelt dollarimärke, et Voilà need ära tunneks)
tehe_1 = r"$$f(x) = a x^2 + b x + c$$"
tehe_2 = r"$$E = m c^2$$"
tehe_3 = r"$$\frac{-b \pm \sqrt{b^2 - 4ac}}{2a}$$"
tehe_4 = r"$$\sum_{i=1}^{n} i = \frac{n(n+1)}{2}$$"
tehe_5 = r"$$\int_{a}^{b} x^2 \,dx = \left[ \frac{x^3}{3} \right]_a^b$$"

# Paneme kõik valemid ühte teksti kokku
kogu_tekst = f"{tehe_1}\n{tehe_2}\n{tehe_3}\n{tehe_4}\n{tehe_5}"

# Kasutame ipywidgets Math vidinat, mida Voilà toetab ametlikult
valemite_liides = widgets.Math(value=kogu_tekst)

# Kuvame liidese
display(valemite_liides)
```
